```python
ticket_details = []
tickets_df = pd.read_csv(os.path.join("../data", "tickets.csv"))
ticket_ids = tickets_df["id"].tolist()

for ticket_id in ticket_ids:
    if ticket_id >= 3040 and ticket_id <= 4218 :
        ticket_details.append(get_ticket_details(ticket_id))
        print(f"ticket_id: {ticket_id}")
        
# Create a new DataFrame from the ticket details
ticket_details_df = pd.DataFrame(ticket_details)
ticket_details_df.to_csv('../data/ticket_details_between_3k_4k.csv', index=False)

```
